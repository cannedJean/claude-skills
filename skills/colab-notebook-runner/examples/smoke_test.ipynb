```python
import subprocess
print(subprocess.run(['nvidia-smi','-L'], capture_output=True, text=True).stdout or 'no gpu')
```

```python
import numpy as np, json
for i in range(3):
    time.sleep(1); print('epoch', i, 'loss', round(float(np.random.rand()), 4))
np.save(f'{OUT}/weights.npy', np.random.rand(256, 256))
json.dump({'acc': 0.93, 'host': platform.node()}, open(f'{OUT}/metrics.json', 'w'))
print(os.listdir(OUT))
```
